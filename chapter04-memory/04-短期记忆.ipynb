### 基于外部存储介质的持久化器

```python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.postgres import PostgresSaver
from llm.my_llm import model_tool
from rich import print as rprint
DB_URL = "postgresql://esired:root@127.0.0.1:5432/langchain_db?sslmode=disable"
with PostgresSaver.from_conn_string(DB_URL) as checkpoint:
    # 初始化postgresql数据库
    checkpoint.setup()

    agent=create_agent(
        model=model_tool,
        checkpointer=checkpoint,
    )

    config={
        'configurable':{
            'thread_id':'1'
        }
    }

    response1=agent.invoke({
        'messages':[
            HumanMessage('你是开源模型？')
        ]
    }, config)


    rprint(response1)
    # for msg in response1['messages']:
    #     msg.pretty_print()
    last_ai_msg = response1['messages'][-1]  # 最后一条消息即为本次模型回复
    print(last_ai_msg)
```
